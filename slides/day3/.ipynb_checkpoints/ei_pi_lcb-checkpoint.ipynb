```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm
from matplotlib.colors import LinearSegmentedColormap

# 1. Distribution parameters for this specific candidate point x
mu = 4.0       # Predicted mean
sigma = 1.5    # Predicted uncertainty (standard deviation)
f_star = 2.8   # Current best minimum threshold
kappa = 2.0    # Exploration factor for LCB
cmap_ei = LinearSegmentedColormap.from_list('ei_gradient', ['#e2f0d9', '#1f4e79'])

# 2. Generate normal distribution curve data
y = np.linspace(mu - 4*sigma, mu + 4*sigma, 1000)
pdf = norm.pdf(y, mu, sigma)

# Calculate the exact LCB point on the horizontal axis
lcb_point = mu - kappa * sigma

# 3. Setup Plot
fig, ax = plt.subplots(figsize=(11, 7))
ax.plot(y, pdf, 'b-', label=r'Posterior Predictive PDF $p(y)$ at point $x$', lw=2.5)

# Threshold line (used by PI and EI)
ax.axvline(f_star, color='red', linestyle='--', lw=2.5, label=r'Current Best Min ($f^*$)')

# 4. Shade the PI / EI region with a gradient
y_imp = y[y <= f_star]
pdf_imp = pdf[y <= f_star]
cmap = LinearSegmentedColormap.from_list('ei_gradient', ['#e2f0d9', '#1f4e79'])

for i in range(len(y_imp) - 1):
    improvement_depth = (f_star - y_imp[i]) / (4 * sigma)
    color = cmap(min(improvement_depth * 2.5, 1.0))
    ax.fill_between(y_imp[i:i+2], 0, pdf_imp[i:i+2], color=color, alpha=0.75)

# 5. Add the LCB Landmark (Point Boundary)
ax.axvline(lcb_point, color='purple', linestyle=':', lw=3, label=r'LCB Point ($\mu - \kappa\sigma$)')
ax.plot(lcb_point, 0, 'o', color='purple', markersize=10, label='LCB Value')

# 6. Annotations to Differentiate All Three
# PI annotation
ax.text(f_star - 0.2, max(pdf)*0.15, r'$\mathbf{PI}$: Measures this total' + '\n' + r'shaded Area ($\int_{-\infty}^{f^*} p(y)dy$)', 
        color='green', fontsize=10, ha='right', weight='bold', bbox=dict(facecolor='white', alpha=0.85, boxstyle='round,pad=0.4'))

# EI annotation
ax.annotate(r'$\mathbf{EI}$: Weights this same area by' + '\n' + r'distance from $f^*$ ($f^* - y$).' + '\n' + r'Rewards a deeper tail (darker blue).', 
            xy=(f_star - 1.8, max(pdf)*0.04), 
            xytext=(f_star - 3.2, max(pdf)*0.4),
            arrowprops=dict(facecolor='#1f4e79', shrink=0.08, width=1.5, headwidth=7),
            fontsize=10, color='#1f4e79', weight='bold', bbox=dict(facecolor='white', alpha=0.85, boxstyle='round,pad=0.4'))

# LCB annotation
ax.annotate(r'$\mathbf{LCB}$ Boundary point:' + '\n' + r'Pushes $\kappa\sigma$ away from $\mu$.' + '\n' + r'Smaller values on axis = Better utility!', 
            xy=(lcb_point, max(pdf)*0.02), 
            xytext=(lcb_point - 1.5, max(pdf)*0.7),
            arrowprops=dict(facecolor='purple', shrink=0.08, width=1.5, headwidth=7),
            fontsize=10, color='purple', weight='bold', bbox=dict(facecolor='white', alpha=0.85, boxstyle='round,pad=0.4'))

# Mean and Sigma Standard markers
ax.text(mu, max(pdf)*0.4, r'Mean $\mu$', color='blue', fontsize=11, ha='center')
ax.plot([mu - sigma, mu], [max(pdf)*0.25, max(pdf)*0.25], 'b--', lw=1.5)
ax.text(mu - sigma/2, max(pdf)*0.28, r'Uncertainty $\sigma$', color='blue', fontsize=10, ha='center')

# Graph Finishes
# ax.set_title('Univariate Breakdown: Geometric Meaning of PI, EI, and LCB', fontsize=13, weight='bold', pad=15)
ax.set_title('PI, EI, and LCB', fontsize=13, weight='bold', pad=15)
ax.set_xlabel(r'Objective Variable outcomes ($y$)', fontsize=11)
ax.set_ylabel('Probability Density', fontsize=11)
ax.grid(True, alpha=0.2)

# Custom legend configuration
from matplotlib.patches import Patch
from matplotlib.lines import Line2D
legend_elements = [
    Line2D([0], [0], color='blue', lw=2, label=r'Prediction Curve $p(y)$'),
    Line2D([0], [0], color='red', linestyle='--', lw=2, label=r'Current Baseline ($f^*$)'),
    Patch(facecolor='#e2f0d9', edgecolor='#1f4e79', label='PI / EI Improvement Horizon Zone'),
    Line2D([0], [0], color='purple', linestyle=':', lw=3, marker='o', markersize=8, label=r'LCB Location ($\kappa = 2$)')
]
ax.legend(handles=legend_elements, loc='upper right', fontsize=10)

plt.tight_layout()
plt.savefig('unified_breakdown.pdf', format='pdf', bbox_inches='tight')
plt.show()

# =====================================================================
# PLOT 2: Standalone Probability of Improvement (PI) Plot
# =====================================================================
fig_pi, ax_pi = plt.subplots(figsize=(9, 5.5))
ax_pi.plot(y, pdf, 'b-', lw=2.5)
ax_pi.axvline(f_star, color='red', linestyle='--', lw=2.5, label=r'Current Best Min ($f^*$)')

# Flat uniform color fill representing strict probability mass
ax_pi.fill_between(y_imp, 0, pdf_imp, color='#99cc99', alpha=0.7, label='Improvement Zone')

ax_pi.text(f_star - 0.2, max(pdf)*0.4, r'$\mathbf{PI(x) = P(Y < f^*)}$' + '\n' + 'Measures the exact size of this unweighted area.', 
           color='darkgreen', fontsize=11, ha='right', weight='bold', bbox=dict(facecolor='white', alpha=0.9, boxstyle='round,pad=0.5'))

ax_pi.set_title('Probability of Improvement (PI)', fontsize=13, weight='bold', pad=15)
ax_pi.set_xlabel(r'$y$')
ax_pi.set_ylabel('Probability Density')
ax_pi.legend(loc='upper right')
ax_pi.grid(True, alpha=0.2)
plt.tight_layout()
plt.savefig('pi_standalone.pdf', format='pdf', bbox_inches='tight')
plt.show()


# =====================================================================
# PLOT 3: Standalone Expected Improvement (EI) Plot
# =====================================================================
fig_ei, ax_ei = plt.subplots(figsize=(9, 5.5))
ax_ei.plot(y, pdf, 'b-', lw=2.5)
ax_ei.axvline(f_star, color='red', linestyle='--', lw=2.5, label=r'Current Best Min ($f^*$)')

# Gradient shift to show distance amplification weighting down the tail
for i in range(len(y_imp) - 1):
    improvement_depth = (f_star - y_imp[i]) / (4 * sigma)
    color = cmap_ei(min(improvement_depth * 2.5, 1.0))
    ax_ei.fill_between(y_imp[i:i+2], 0, pdf_imp[i:i+2], color=color, alpha=0.85)

ax_ei.text(f_star - 0.3, max(pdf)*0.4, r'$\mathbf{EI(x) = \int_{-\infty}^{f^*} (f^* - y) p(y) dy}$' + '\n' + 'Weights the area by distance from threshold.\nRewards deeper tail risks (dark blue).', 
           color='#1f4e79', fontsize=11, ha='right', weight='bold', bbox=dict(facecolor='white', alpha=0.9, boxstyle='round,pad=0.5'))

ax_ei.set_title('Expected Improvement (EI)', fontsize=13, weight='bold', pad=15)
ax_ei.set_xlabel(r'$y$')
ax_ei.set_ylabel('Probability Density')
ax_ei.grid(True, alpha=0.2)
plt.tight_layout()
plt.savefig('ei_standalone.pdf', format='pdf', bbox_inches='tight')
plt.show()


# =====================================================================
# PLOT 4: Standalone Lower Confidence Bound (LCB) Plot
# =====================================================================
fig_lcb, ax_lcb = plt.subplots(figsize=(9, 5.5))
ax_lcb.plot(y, pdf, 'b-', lw=2.5)

# Highlight mean and show horizontal shift vector out to the boundary marker
ax_lcb.axvline(mu, color='blue', linestyle=':', alpha=0.6, lw=1.5)
ax_lcb.axvline(lcb_point, color='purple', linestyle=':', lw=2.5)
ax_lcb.plot(lcb_point, 0, 'o', color='purple', markersize=10, label=r'LCB Value')

# Arrow showing the scalar subtraction vector down to the boundary edge
ax_lcb.annotate('', xy=(lcb_point, max(pdf)*0.1), xytext=(mu, max(pdf)*0.1),
                arrowprops=dict(arrowstyle="<->", color='purple', lw=2))
ax_lcb.text(mu - (kappa*sigma)/2, max(pdf)*0.13, r'$\kappa \cdot \sigma$', color='purple', fontsize=12, ha='center', weight='bold')

ax_lcb.text(lcb_point - 0.3, max(pdf)*0.4, r'$\mathbf{LCB(x) = \mu(x) - \kappa\sigma(x)}$' + '\n' + 'Ignores $f^*$ completely. Identifies a fixed safety\nboundary point along the horizontal domain axis.', 
           color='purple', fontsize=11, ha='right', weight='bold', bbox=dict(facecolor='white', alpha=0.9, boxstyle='round,pad=0.5'))

ax_lcb.set_title(r'Lower Confidence Bound (LCB)', fontsize=13, weight='bold', pad=15)
ax_lcb.set_xlabel(r'$y$')
ax_lcb.set_ylabel('Probability Density')
ax_lcb.grid(True, alpha=0.2)
plt.tight_layout()
plt.savefig('lcb_standalone.pdf', format='pdf', bbox_inches='tight')
plt.show()
```
